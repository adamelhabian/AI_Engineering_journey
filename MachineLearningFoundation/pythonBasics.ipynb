```python
class Dog(Animal):
```

```python
class Dog(Animal):
```

<summary>The Four Pillars of OOP</summary>

The **four pillars of Object-Oriented Programming (OOP)** are:

2. **Abstraction**

---

## 1. Encapsulation

**Encapsulation** means **bundling data and the methods that work on that data inside one class**, while controlling how that data is accessed or changed.

### Example

```python
class BankAccount:
    def __init__(self, balance):
        self.__balance = balance

    def deposit(self, amount):
        self.__balance += amount

    def get_balance(self):
        return self.__balance
```

Here:

* `__balance` → hidden/private attribute
* `deposit()` → controls how the balance changes
* `get_balance()` → provides controlled access to the balance

Instead of directly changing:

```python
account.__balance = -100000
```

we use methods that control the data.

### Easy idea

> **Encapsulation = Protect the data + control access to it.**

---

## 2. Abstraction

**Abstraction** means **hiding unnecessary implementation details** and showing only what the user needs.

For example, when you use:

```python
car.start()
```

you don't need to know all the internal steps required to start the engine.

You only need to know:

> "Call `start()` to start the car."

### Python example

```python
from abc import ABC, abstractmethod

class Animal(ABC):

    @abstractmethod
    def make_sound(self):
        pass
```

The class says:

> Every animal must have `make_sound()`.

But it doesn't specify exactly how every animal makes its sound.

```python
class Dog(Animal):
    def make_sound(self):
        print("Bark")

class Cat(Animal):
    def make_sound(self):
        print("Meow")
```

### Easy idea

> **Abstraction = Hide the complicated details + show only the important part.**

---

## 3. Inheritance

**Inheritance** means creating a new class from an existing class.

The new class can **reuse** attributes and methods from the parent class.

```python
class Animal:
    def eat(self):
        print("Eating")

class Dog(Animal):
    def bark(self):
        print("Barking")
```

Now `Dog` inherits from `Animal`.

```python
dog = Dog()

dog.eat()    # Eating
dog.bark()  # Barking
```

`Dog` didn't define `eat()`, but it can still use it because it inherited it from `Animal`.

### Terminology

```text
Animal → Parent / Base class
Dog    → Child / Derived class
```

### 3. Private members

# 4-Abstraction
Abstraction is the process of hiding implementation details and exposing only the essential functionality to the user. It is used to hide the implementation details from the user and expose only necessary parts, making the code simpler and easier to interact with.

### Abstract Base Class 
Abstract Base Class
Abstract Base Class (ABC) is used to achieve data abstraction by defining a common interface for its subclasses. It cannot be instantiated directly and serves as a blueprint for other classes.

Abstract classes are created using abc module and @abstractmethod decorator, allowing developers to enforce method implementation in subclasses while hiding complex internal logic.

````python
from abc import ABC , abstractmethod

# define a clear and enforced structure for subclasses 
class Greet(ABC):

    @abstractmethod
    def say_hello(self):
        pass  # Abstract method

class English(Greet):
    def say_hello(self):
        return "Hello!"

g = English()
print(g.say_hello())
````

### 1. Abstract Method
Abstract methods are method declarations without a body defined inside an abstract class. They act as placeholders that force subclasses to provide their own specific implementation, ensuring consistent structure across derived classes

````python
from abc import ABC, abstractmethod
class Animal(ABC):
    @abstractmethod
    def make_sound(self):
        pass  # Abstract method, no implementation here
````

### 2. Concrete Method
Concrete methods are fully implemented methods within an abstract class. Subclasses can inherit and use them directly, promoting code reuse without needing to redefine common functionality.






````python
from abc import ABC, abstractmethod

class Animal(ABC):
    @abstractmethod
    def make_sound(self):
        pass

    def move(self):
        return "Moving"

class Dog(Animal):
    def make_sound(self):
        return "Bark"

dog = Dog()
print(dog.move())
````

### 3. Abstract Properties
Abstract properties work like abstract methods but are used for properties. These properties are declared with @property decorator and marked as abstract using @abstractmethod. Subclasses must implement these properties

````python
from abc import ABC, abstractmethod

class Animal(ABC):
    @property
    @abstractmethod
    def species(self):
        pass  # Abstract property, must be implemented by subclasses

class Dog(Animal):
    @property
    def species(self):
        return "Canine"

# Instantiate the concrete subclass
dog = Dog()
print(dog.species)
````

Abstract classes cannot be instantiated directly. This is because they contain one or more abstract methods or properties that lack implementations. Attempting to instantiate an abstract class results in a TypeError.

````python
from abc import ABC, abstractmethod

class Animal(ABC):
    @abstractmethod
    def make_sound(self):
        pass

#animal = Animal()
````

````python
class Temp:
    @staticmethod
    def to_fahrenheit(c):
        return (c * 9/5) + 32

print(Temp.to_fahrenheit(25))
````

````python
class Person:
    @staticmethod
    def is_adult(age):
        return age >= 18

print(Person.is_adult(16))
print(Person.is_adult(21))
````

<details>
<summary><code>__str__()</code> Function</summary>

`__str__()` is a **special method** that defines how an object is represented as a **string**.

It is automatically called by `print()` and `str()`.

### Example

```python
class Person:
    def __init__(self, name):
        self.name = name

    def __str__(self):
        return self.name


person = Person("Adam")

print(person)
```

Output:

```text
Adam
```

Without `__str__()`, Python usually prints something like:

```text
<__main__.Person object at 0x...>
```

### Important Rule

`__str__()` **must return a string**.

```python
def __str__(self):
    return "Hello"
```

### Easy Rule

> **`__str__()` → controls what you see when you `print(object)`.**

</details>


````python
class Dog:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def __str__(self):
        return f"{self.name} is {self.age} years old."
dog1 = Dog("Buddy", 3)
dog2 = Dog("Charlie", 5)

print(dog1)  
print(dog2)
````

<details>
<summary>Metaclass</summary>

A **metaclass** is a **class that creates other classes**.

Think of it like:

```text
Object → created by a Class
Class  → created by a Metaclass
```

### Simple Example

Normally, we create a class:

```python
class Person:
    pass
```

Python internally creates `Person` using a metaclass.

The default metaclass is:

```python
type
```

So:

```python
print(type(Person))
```

Output:

```text
<class 'type'>
```

This means:

> `Person` is an object, and its class is `type`.

### `type` as a Metaclass

You can also create a class using `type()`:

```python
Person = type(
    "Person",
    (),
    {}
)
```

Now `Person` is a class.

```python
person = Person()
```

### Custom Metaclass

You can create your own metaclass:

```python
class MyMeta(type):
    pass


class Person(metaclass=MyMeta):
    pass
```

Here:

```text
MyMeta
   ↓ creates
Person
   ↓ creates
person
```

### Why Use Metaclasses?

Metaclasses allow you to **control how classes are created**.

They can be used to:

* Automatically add attributes or methods.
* Validate classes when they are created.
* Enforce rules on classes.
* Modify class behavior.

### Easy Rule

> **Class → creates objects**

> **Metaclass → creates classes**

</details>


<details>
<summary><code>__new__()</code> vs <code>__init__()</code></summary>

|              | `__new__()`                 | `__init__()`              |
| ------------ | --------------------------- | ------------------------- |
| **Purpose**  | Creates the object          | Initializes the object    |
| **Runs**     | First                       | After `__new__()`         |
| **Receives** | `cls`                       | `self`                    |
| **Returns**  | Must return an object       | Usually returns `None`    |
| **Used for** | Controlling object creation | Setting object attributes |
| **Usage**    | Less common                 | Very common               |

### Example

```python
class Person:

    def __new__(cls, name):
        print("__new__")
        return super().__new__(cls)

    def __init__(self, name):
        print("__init__")
        self.name = name


p = Person("Adam")
```

Output:

```text
__new__
__init__
```

### What happens?

When you write:

```python
p = Person("Adam")
```

Python does:

```text
1. __new__() → creates the object
        ↓
2. __init__() → initializes the object
```

### Easy Rule

> `__new__()` → **Create the object**

> `__init__()` → **Initialize the object**

</details>


# Self as Default Argument

In Python, when defining methods inside a class, first parameter is always **self**. It is not a keyword, but a naming convention that plays a key role in Python’s object-oriented programming. The self parameter represents instance of the class itself, allowing you to access and modify its attributes and methods.

````python
# Self is a convention not python keyword so i can write any thing instead of self and play the same role 
class Car:
    def __init__(this, brand, model):
        this.brand = brand  # Set instance attribute
        this.model = model  # Set instance attribute

    def display(this):
        print(this.__class__)
        return this.brand, this.model
        

# Create an instance of Car
car1 = Car("Toyota", "Corolla")

# Call the display method
print(car1.display())
````

### `iter()`

`iter()` converts an **iterable** (like a list, tuple, or string) into an **iterator**.

```python
numbers = [10, 20, 30]

it = iter(numbers)
```

Now `it` is an iterator.

```text
List → iter() → Iterator
```

---

### `next()`

`next()` gets the **next item** from an iterator.

```python
print(next(it))  # 10
print(next(it))  # 20
print(next(it))  # 30
```

When there are no more items:

```python
next(it)
```

Python raises:

```text
StopIteration
```

---

### How They Work Together

```python
numbers = [10, 20, 30]

it = iter(numbers)

print(next(it))  # 10
print(next(it))  # 20
print(next(it))  # 30
```

Think of it as:

```text
numbers
   ↓
iter()
   ↓
Iterator
   ↓
next() → 10
   ↓
next() → 20
   ↓
next() → 30
   ↓
next() → StopIteration
```

### Easy Rule

> **`iter()` → Creates an iterator**

> **`next()` → Gets the next item**

</details>
