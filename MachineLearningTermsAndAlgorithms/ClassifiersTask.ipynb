## Model Accuracy Comparison Task

Compare the accuracy of Logistic Regression, Gaussian Naive Bayes, KNN, Linear SVM, and RBF SVM.

```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df=pd.read_csv('CSVdata/Teen_Mental_Health_Dataset.csv')
df.head()
```

```python
print(df.shape)
```

```python
x=df[['age','daily_social_media_hours','sleep_hours','screen_time_before_sleep','academic_performance','physical_activity','stress_level','anxiety_level','addiction_level']]
y=df['depression_label']
```

```python
print(x.isnull().sum())
print(y.isnull().sum())
```

```python
from sklearn.model_selection import train_test_split

x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.3,random_state=42,stratify=y)
```

```python
print(len(x_train))
print(len(x_test))
```

```python
from sklearn.preprocessing import StandardScaler

scaler=StandardScaler()
x_train=scaler.fit_transform(x_train)
x_test=scaler.transform(x_test)
```

```python
from sklearn.linear_model import LogisticRegression

model=LogisticRegression()
model.fit(x_train,y_train)
```

```python
y_pred_logistic=model.predict(x_test)
y_prob=model.predict_proba(x_test)
```

```python
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    classification_report,
    confusion_matrix,
)

print("Accuracy :", accuracy_score(y_test, y_pred_logistic))
print("Precision:", precision_score(y_test, y_pred_logistic))
print("Recall   :", recall_score(y_test, y_pred_logistic))
print("F1 Score :", f1_score(y_test, y_pred_logistic))

print("\nClassification Report")
print(classification_report(y_test, y_pred_logistic))
```

```python
from sklearn.svm import SVC

# Linear SVM
linear_model = SVC(kernel="linear", random_state=42)

# RBF SVM
rbf_model = SVC(kernel="rbf", random_state=42)

linear_model.fit(x_train,y_train)
rbf_model.fit(x_train,y_train)
```

```python
y_pred_linear=linear_model.predict(x_test)
y_pred_rbf=rbf_model.predict(x_test)
```

```python
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    classification_report,
)

print(" Linear SVM :")
print("Accuracy :", round(accuracy_score(y_test, y_pred_linear), 3))
print("Precision:", round(precision_score(y_test, y_pred_linear), 3))
print("Recall   :", round(recall_score(y_test, y_pred_linear), 3))
print("F1 Score :", round(f1_score(y_test, y_pred_linear), 3))
print("\nClassification Report:")
print(classification_report(y_test, y_pred_linear))


print("\nRBF SVM :")
print("Accuracy :", round(accuracy_score(y_test, y_pred_rbf), 3))
print("Precision:", round(precision_score(y_test, y_pred_rbf), 3))
print("Recall   :", round(recall_score(y_test, y_pred_rbf), 3))
print("F1 Score :", round(f1_score(y_test, y_pred_rbf), 3))
print("\nClassification Report:")
print(classification_report(y_test, y_pred_rbf))
```

```python
from sklearn.naive_bayes import GaussianNB

model=GaussianNB()
model.fit(x_train,y_train)
```

```python
y_pred_nb=model.predict(x_test)
```

```python
print("Accuracy:", round(accuracy_score(y_test, y_pred_nb), 3))
#pos_label : Which class should I consider the "positive" class when calculating precision, recall, and F1?
print("Precision:", round(precision_score(y_test, y_pred_nb), 3))
print("Recall:", round(recall_score(y_test, y_pred_nb), 3))
print("F1:", round(f1_score(y_test, y_pred_nb ), 3))
print("\nClassification Report:")
print(classification_report(y_test, y_pred_nb))
```

```python
from sklearn.neighbors import KNeighborsClassifier

classifier=KNeighborsClassifier(n_neighbors=5)
classifier.fit(x_train,y_train)
```

```python
y_pred_KNN=classifier.predict(x_test)
```

```python
print("Accuracy:", round(accuracy_score(y_test, y_pred_KNN), 3))
#pos_label : Which class should I consider the "positive" class when calculating precision, recall, and F1?
print("Precision:", round(precision_score(y_test, y_pred_KNN), 3))
print("Recall:", round(recall_score(y_test, y_pred_KNN), 3))
print("F1:", round(f1_score(y_test, y_pred_KNN ), 3))
print("\nClassification Report:")
print(classification_report(y_test, y_pred_KNN))
```

```python
from sklearn.metrics import accuracy_score
import pandas as pd

comparison = pd.DataFrame({
    "Model": [
        "Logistic Regression",
        "Gaussian Naive Bayes",
        "KNN",
        "Linear SVM",
        "RBF SVM"
    ],
    "Accuracy": [
        round(accuracy_score(y_test, y_pred_logistic), 3),
        round(accuracy_score(y_test, y_pred_nb), 3),
        round(accuracy_score(y_test, y_pred_KNN), 3),
        round(accuracy_score(y_test, y_pred_linear), 3),
        round(accuracy_score(y_test, y_pred_rbf), 3)
    ]
})

comparison = comparison.sort_values(by="Accuracy", ascending=False)

print(comparison)
```
