# K-Nearest Neighbors (KNN) Regression

<details>
<summary><strong>What is KNN Regression?</strong></summary>

K-Nearest Neighbors (KNN) Regression is a **supervised machine learning algorithm** used to predict **continuous numerical values**.

Unlike KNN Classification, which predicts a class label, KNN Regression predicts a number by averaging the target values of the nearest neighbors.

**Goal:** Predict the target value based on the values of the **K nearest training samples**.

</details>

---

<details>
<summary><strong>What is K?</strong></summary>

`K` is the **number of nearest neighbors** the algorithm uses to make a prediction.

Example:

```python
KNeighborsRegressor(n_neighbors=5)
```

The model will use the **5 closest training samples** when predicting a new value.

Choosing the correct value of **K** is important because it affects the model's performance.

</details>

---

<details>
<summary><strong>How Does KNN Regression Work?</strong></summary>

### Step 1
Choose the number of neighbors (**K**).

### Step 2
Calculate the distance between the new sample and every training sample.

### Step 3
Find the **K nearest neighbors**.

### Step 4
Take the target values of those neighbors.

### Step 5
Compute the **mean (average)** of their target values.

### Step 6
Return that average as the predicted value.

</details>

---

<details>
<summary><strong>Example</strong></summary>

Suppose we have the following training data:

| Age | Salary |
|----:|-------:|
| 22 | 30,000 |
| 25 | 35,000 |
| 28 | 40,000 |
| 35 | 60,000 |
| 40 | 70,000 |

We want to predict the salary for:

```text
Age = 27
```

If:

```text
K = 3
```

Nearest neighbors:

- Age = 25 → Salary = 35,000
- Age = 28 → Salary = 40,000
- Age = 22 → Salary = 30,000

Prediction:

$$
\frac{35000+40000+30000}{3}=35000
$$

Predicted Salary:

```text
35,000
```

</details>

---

<details>
<summary><strong>Distance Formula</strong></summary>

KNN uses **Euclidean Distance** by default.

$$
d=\sqrt{(x_1-x_2)^2+(y_1-y_2)^2}
$$

For datasets with more than two features:

$$
d=\sqrt{\sum_{i=1}^{n}(x_i-y_i)^2}
$$

The nearest neighbors are selected based on the smallest distance.

</details>

---

<details>
<summary><strong>Training Process</strong></summary>

```python
from sklearn.neighbors import KNeighborsRegressor

model = KNeighborsRegressor(
    n_neighbors=5
)

model.fit(X_train, y_train)
```

Unlike many machine learning algorithms, KNN has **no actual training phase**.

During `fit()`, the model simply stores the training dataset.

</details>

---

<details>
<summary><strong>Prediction Process</strong></summary>

```python
prediction = model.predict(X_test)
```

During prediction, KNN:

- Computes the distance to every training sample.
- Finds the K nearest neighbors.
- Averages their target values.
- Returns the predicted numerical value.

</details>

---

<details>
<summary><strong>Choosing the Best K</strong></summary>

A small value of **K**:

- Sensitive to noise.
- Can overfit the data.
- High variance.

A large value of **K**:

- Produces smoother predictions.
- May underfit the data.
- Higher bias.

The best value of **K** is usually selected using **Cross Validation**.

</details>

---

<details>
<summary><strong>Feature Scaling</strong></summary>

KNN is a **distance-based algorithm**.

Features should be scaled before training.

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)
```

Without scaling, features with larger values dominate the distance calculation.

</details>

---

<details>
<summary><strong>Advantages</strong></summary>

- Simple and easy to understand.
- No training phase.
- Works well for small datasets.
- Can model non-linear relationships.
- Easy to implement.

</details>

---

<details>
<summary><strong>Disadvantages</strong></summary>

- Prediction becomes slow for large datasets.
- Requires feature scaling.
- Sensitive to noisy data and outliers.
- Choosing the right value of **K** can be difficult.
- Requires storing the entire training dataset.

</details>

---

<details>
<summary><strong>Applications</strong></summary>

Common uses of KNN Regression include:

- House Price Prediction
- Salary Prediction
- Sales Forecasting
- Medical Measurements
- Energy Consumption Prediction
- Demand Forecasting

</details>

---

<details>
<summary><strong>KNN Regression vs KNN Classification</strong></summary>

| KNN Regression | KNN Classification |
|---------------|--------------------|
| Predicts a numerical value | Predicts a class label |
| Uses the mean of neighbors | Uses majority voting |
| Output is continuous | Output is categorical |
| Used for regression problems | Used for classification problems |

</details>

<details>
<summary><strong>Summary</strong></summary>

- KNN Regression is a **supervised learning** algorithm.
- Used to predict **continuous numerical values**.
- Chooses the **K nearest neighbors**.
- Uses **Euclidean Distance** by default.
- Predicts by averaging the neighbors' target values.
- Requires **feature scaling**.
- The value of **K** greatly affects performance.
- Has no real training phase; it stores the training data and performs computations during prediction.

</details>

````python
from sklearn.neighbors import KNeighborsRegressor
from sklearn.datasets import make_regression
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_squared_error,r2_score
import matplotlib.pyplot as plt

# generates a synthetic regression dataset follow standard normally distribution
# x will be features matrix , y is the target value vector
x,y=make_regression(n_samples=200,n_features=2,random_state=42)

print("Features of first 10 samples : \n",x[:10])
print()
print("Target Values of first 10 samples : \n",y[:10])
````

````python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
print(len(x_train))
print(len(x_test))
````

````python
knn_regressor=KNeighborsRegressor(n_neighbors=5)
knn_regressor.fit(x_train,y_train)
````

````python
y_pred=knn_regressor.predict(x_test)
print("y_predition of x_tested :\n",y_pred)
````

````python
# Model Evalutaion 
MSE=mean_squared_error(y_test,y_pred)
r2=r2_score(y_test,y_pred)
print("Mean Squared Error :",MSE)
print("R-Squared :",r2)
````
