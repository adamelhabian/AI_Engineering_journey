# SVM Algorithms Overview

<details>
<summary><strong>1. Linear SVC (SVC with Linear Kernel)</strong></summary>

### Description

`SVC(kernel='linear')` is a Support Vector Machine classifier that separates classes using a **straight-line decision boundary (hyperplane)**.

### Characteristics

- Classification algorithm
- Uses a **linear kernel**
- Best for linearly separable data
- Supports support vectors
- Can calculate probabilities (if enabled)

### Example

```python
from sklearn.svm import SVC

model = SVC(kernel='linear', C=1.0)
```

### Best For

- Small to medium datasets
- Data that can be separated with a straight line

</details>

---

<details>
<summary><strong>2. LinearSVC</strong></summary>

### Description

`LinearSVC` is another implementation of a **linear Support Vector Machine** optimized for speed.

### Characteristics

- Classification algorithm
- Uses only a **linear decision boundary**
- Faster than `SVC(kernel='linear')` on large datasets
- Does **not** support non-linear kernels

### Example

```python
from sklearn.svm import LinearSVC

model = LinearSVC(C=1.0, max_iter=1000)
```

### Best For

- Large datasets
- High-dimensional data (many features)

</details>

---

<details>
<summary><strong>3. SVC with RBF Kernel</strong></summary>

### Description

`SVC(kernel='rbf')` creates a **flexible curved decision boundary** using the **Radial Basis Function (RBF)** kernel.

### Characteristics

- Classification algorithm
- Handles complex, non-linear data
- Most commonly used SVM kernel
- Controlled by **C** and **gamma**

### Example

```python
from sklearn.svm import SVC

model = SVC(
    kernel='rbf',
    C=1.0,
    gamma=0.7
)
```

### Best For

- Non-linear datasets
- Complex classification problems

</details>

---

<details>
<summary><strong>4. SVC with Polynomial Kernel</strong></summary>

### Description

`SVC(kernel='poly')` creates a **polynomial-shaped decision boundary**.

### Characteristics

- Classification algorithm
- Suitable for polynomial relationships
- Controlled by **degree**, **C**, and **gamma**

### Example

```python
from sklearn.svm import SVC

model = SVC(
    kernel='poly',
    degree=3,
    C=1.0,
    gamma='auto'
)
```

### Best For

- Data with polynomial relationships
- When a simple curved boundary is sufficient

</details>

---

# Parameter Summary

| Parameter | Description |
|------------|-------------|
| `kernel` | Type of decision boundary (`linear`, `rbf`, `poly`) |
| `C` | Controls the balance between a larger margin and fewer training errors |
| `gamma` | Controls how flexible the decision boundary is (used with `rbf` and `poly`) |
| `degree` | Degree of the polynomial kernel |
| `max_iter` | Maximum number of training iterations |
| `random_state` | Makes results reproducible when randomness is involved |

---

# Algorithm Comparison

| Algorithm | Decision Boundary | Speed | Best For |
|------------|-------------------|-------|----------|
| **SVC (Linear)** | Straight line | Fast | Small/medium linearly separable datasets |
| **LinearSVC** | Straight line | Very Fast | Large datasets with many features |
| **SVC (RBF)** | Flexible curved boundary | Medium | Most non-linear datasets |
| **SVC (Polynomial)** | Polynomial curve | Medium | Polynomial relationships |

---

# Which One Should You Choose?

| If... | Use |
|--------|-----|
| Data is linearly separable | `SVC(kernel='linear')` |
| Dataset is very large | `LinearSVC` |
| Data is complex and non-linear | `SVC(kernel='rbf')` |
| Data follows a polynomial pattern | `SVC(kernel='poly')` |

---

# Quick Summary

- **SVC (Linear):** Standard SVM with a straight-line decision boundary.
- **LinearSVC:** Faster implementation of linear SVM for large datasets.
- **SVC (RBF):** Most popular non-linear SVM using a flexible curved boundary.
- **SVC (Polynomial):** Uses a polynomial curve to separate classes.
- **C:** Controls model strictness and the trade-off between margin size and training errors.
- **Gamma:** Controls the flexibility of non-linear decision boundaries.
- **Degree:** Controls the complexity of the polynomial kernel.

````python
# we create an instance of SVM and fit out data. We do not scale our
# data since we want to plot the support vectors
C = 1.0  # SVM regularization parameter , Control how strict the model is , small c simpler model , large c tries harder to classify every training sample correctly
models = (
    svm.SVC(kernel="linear", C=C),
    svm.LinearSVC(C=C, max_iter=10000, dual="auto"),# this is another implementaion of linear SVM ,
     # dual='auto' mean tell sckit-learn to automatic choose the best optimization method depend on the dataset
    svm.SVC(kernel="rbf", gamma=0.7, C=C), # RBF Can Create curved decision boundaries , gamma control how flexible the curved boundary is
    svm.SVC(kernel="poly", degree=3, gamma="auto", C=C), # create a polynomial shaped boundary
)
models=(clf.fit(X, y) for clf in models)
````

````python
models
````

````python
# title for the plots
titles = (
    "SVC with linear kernel",
    "LinearSVC (linear kernel)",
    "SVC with RBF kernel",
    "SVC with polynomial (degree 3) kernel",
)
````

````python
# Set-up 2x2 grid for plotting.
fig, sub = plt.subplots(2, 2)
plt.subplots_adjust(wspace=0.4, hspace=0.4)

X0, X1 = X[:, 0], X[:, 1]

for clf, title, ax in zip(models, titles, sub.flatten()):
    disp = DecisionBoundaryDisplay.from_estimator(
        clf,
        X,
        response_method="predict",
        cmap=plt.cm.coolwarm,
        alpha=0.8,
        ax=ax,
        xlabel=iris.feature_names[0],
        ylabel=iris.feature_names[1],
    )
    ax.scatter(X0, X1, c=y, cmap=plt.cm.coolwarm, s=20, edgecolors="k")
    ax.set_xticks(())
    ax.set_yticks(())
    ax.set_title(title)

plt.show()
````
