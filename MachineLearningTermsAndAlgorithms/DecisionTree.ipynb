# Decision Tree

<details>
<summary><strong>What is a Decision Tree?</strong></summary>

A **Decision Tree** is a **supervised machine learning algorithm** used for both:

- Classification tasks
- Regression tasks

It learns a sequence of **if-else decision rules** from the training data to make predictions.

The algorithm represents decisions in a tree structure:

```
                 Root Node
                    |
              Feature Decision
             /              \
        Branch              Branch
          |                   |
      Decision            Decision
          |                   |
      Leaf Node          Leaf Node
      Output             Output
```

The goal of a Decision Tree is to split the data into groups where samples inside each group are as similar as possible.

</details>

---

<details>
<summary><strong>Main Components of a Decision Tree</strong></summary>

A Decision Tree consists of:

## 1. Root Node

The first decision point in the tree.

It contains the feature that creates the best split.

Example:

```
              Income > 50k?
```

---

## 2. Internal Node

A node that contains a decision rule.

Example:

```
              Age < 30?
```

---

## 3. Branch

The result of a decision.

Example:

```
Age < 30?

Yes → Left Branch

No → Right Branch
```

---

## 4. Leaf Node

The final prediction.

Example:

```
Prediction:
Customer will buy
```

</details>

---

<details>
<summary><strong>How Does a Decision Tree Work?</strong></summary>

A Decision Tree builds the model by repeatedly splitting the dataset.

At every step, it asks:

> Which feature and threshold create the best separation between the data?

Example:

Dataset:

| Age | Income | Buy |
|---|---|---|
|22|30k|No|
|25|40k|No|
|35|70k|Yes|
|45|90k|Yes|

The algorithm may learn:

```
              Age > 30?
              /       \
            Yes        No
            |           |
          Buy        Not Buy
```

The splitting process continues until:

- The nodes become pure.
- Maximum depth is reached.
- Minimum samples condition is reached.

</details>

---

<details>
<summary><strong>How Does the Algorithm Choose the Best Split?</strong></summary>

The Decision Tree tries many possible splits.

Example:

```
Age < 20

Age < 30

Income < 50k

Experience < 5 years
```

For each split, it measures how well the split separates the data.

The algorithm chooses the split that creates the purest child nodes.

For classification, common split criteria are:

- Gini Impurity
- Entropy
- Information Gain

</details>

---

<details>
<summary><strong>Gini Impurity</strong></summary>

Gini Impurity measures how mixed the classes are inside a node.

Formula:

$$
Gini = 1 - \sum p_i^2
$$

Where:

- $p_i$ is the probability of each class.

Example:

### Pure Node

```
10 samples

10 Cats
0 Dogs
```

The node contains only one class.

```
Gini = 0
```

---

### Mixed Node

```
10 samples

5 Cats
5 Dogs
```

The classes are completely mixed.

Higher Gini value.

The Decision Tree prefers splits that reduce Gini impurity.

</details>

---

<details>
<summary><strong>Entropy and Information Gain</strong></summary>

Entropy measures the uncertainty inside a node.

Formula:

$$
Entropy = -\sum p_i log_2(p_i)
$$

Low entropy:

```
One class dominates
```

High entropy:

```
Classes are mixed
```

Information Gain measures how much uncertainty decreases after a split.

The tree chooses the split with the highest information gain.

</details>

---

<details>
<summary><strong>Decision Tree Classification vs Regression</strong></summary>

## Decision Tree Classification

Used when the target is a category.

Examples:

```
Email → Spam / Not Spam

Disease → Positive / Negative
```

Scikit-learn:

```python
from sklearn.tree import DecisionTreeClassifier

model = DecisionTreeClassifier()
```

The output is a class label.

---

## Decision Tree Regression

Used when the target is a continuous value.

Examples:

```
House Features → House Price

Age + Experience → Salary
```

Scikit-learn:

```python
from sklearn.tree import DecisionTreeRegressor

model = DecisionTreeRegressor()
```

The output is a numerical value.

</details>

---

<details>
<summary><strong>How Does Decision Tree Training Work?</strong></summary>

Example:

Features:

```
Age
Income
Education
```

Target:

```
Buy Product
```

Training process:

### Step 1

Try possible splits:

```
Age < 30

Income < 50k

Education = University
```

---

### Step 2

Calculate impurity after each split.

---

### Step 3

Choose the best split.

Example:

```
Income < 50k
```

becomes the root node.

---

### Step 4

Repeat the same process on the remaining data.

The tree grows until stopping conditions are reached.

</details>

---

<details>
<summary><strong>Training a Decision Tree Using Scikit-Learn</strong></summary>

Example:

```python
from sklearn.tree import DecisionTreeClassifier

model = DecisionTreeClassifier(
    random_state=42
)

model.fit(
    X_train,
    y_train
)
```

During training, the algorithm:

1. Finds the best feature.
2. Finds the best threshold.
3. Creates a split.
4. Repeats the process recursively.

</details>

---

<details>
<summary><strong>Making Predictions</strong></summary>

After training:

```python
y_pred = model.predict(X_test)
```

For a new sample:

```
Income > 50k?

        Yes

        Yes

Prediction:

```

The sample travels through the tree until reaching a leaf node.

</details>

---

<details>
<summary><strong>Controlling Tree Complexity</strong></summary>

Decision Trees can easily overfit because they can memorize training data.

Example:

Deep tree:

```
Many decisions
        |
        |
Very specific rules
        |
Overfitting
```

Important parameters:

---

## max_depth

Controls maximum tree depth.

```python
DecisionTreeClassifier(
    max_depth=5
)
```

Smaller depth:

- Simpler model
- Less overfitting


---

## min_samples_split

Minimum number of samples required to split a node.

```python
min_samples_split=10
```

---

## min_samples_leaf

Minimum samples required in a leaf node.

```python
min_samples_leaf=5
```

</details>

---

<details>
<summary><strong>Feature Importance</strong></summary>

Decision Trees can calculate the importance of each feature.

Example:

```python
model.feature_importances_
```

Output:

```
Age        0.65
Income     0.30
Country    0.05
```

This means:

Age contributed more to the model decisions.

</details>

---

<details>
<summary><strong>Does Decision Tree Require Feature Scaling?</strong></summary>

No.

Decision Trees do **not require feature scaling**.

Why?

Because they do not calculate distances.

They only compare values using thresholds.

Example:

```
Age < 30

Income < 50000
```

Unlike:

- KNN
- K-Means
- SVM

which depend on distance calculations.

</details>

---

<details>
<summary><strong>Advantages</strong></summary>

- Easy to understand and visualize.
- Works with numerical and categorical features.
- Does not require feature scaling.
- Can capture nonlinear relationships.
- Requires less data preprocessing.
- Provides feature importance.

</details>

---

<details>
<summary><strong>Disadvantages</strong></summary>

- Can easily overfit.
- Sensitive to small changes in data.
- Large trees become difficult to interpret.
- Usually less accurate than ensemble methods.

Examples of ensemble methods:

- Random Forest
- Gradient Boosting
- XGBoost

</details>

---

<details>
<summary><strong>Applications</strong></summary>

Decision Trees are used in:

- Medical diagnosis
- Fraud detection
- Credit risk analysis
- Customer churn prediction
- Loan approval systems
- Recommendation systems
- Business decision making

</details>

---

<details>
<summary><strong>Summary</strong></summary>

- Decision Tree is a supervised learning algorithm.
- It learns rules in the form of if-else conditions.
- It can solve classification and regression problems.
- It splits data using criteria like Gini Impurity and Entropy.
- Root nodes represent important decisions.
- Leaf nodes contain final predictions.
- It does not require feature scaling.
- Deep trees may overfit.
- Parameters like `max_depth` control model complexity.

</details>

# Import Required Librairies

````python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier, plot_tree
from sklearn.metrics import (
    accuracy_score,
    confusion_matrix,
    ConfusionMatrixDisplay,
    classification_report
)

print("Libraries imported successfully.")
````

# Load DataSet

````python
iris = load_iris()

df=pd.DataFrame(iris.data,columns=iris.feature_names)
df['target']=iris.target
df['species']=df['target'].map(dict(enumerate(iris.target_names)))
df.head()

````

# Explore the Dataset

````python
print("Dataset shape:", df.shape)

print("\nDataset information:")
df.info()

print("\nSummary statistics:")
display(df.describe())

print("\nClass distribution:")
display(df["species"].value_counts())
````

# Separate Features and Target

````python
X = df[iris.feature_names]
y = df["target"]

print("X shape:", X.shape)
print("y shape:", y.shape)
````

# Split the Dataset

````python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training samples:", X_train.shape[0])
print("Testing samples:", X_test.shape[0])
````

# Create Decision Tree Model

````python
# Criterion -> How should you decide which split is the best? Use Gini Impurity to measure how good a split is
model=DecisionTreeClassifier(criterion="gini",max_depth=3,random_state=42)
````

# Train the Model

````python
model.fit(X_train, y_train)
print("Model trained successfully.")
````

# Make Prediction

````python
y_pred = model.predict(X_test)

comparison = pd.DataFrame({
    "Actual": y_test.values,
    "Predicted": y_pred
})

comparison['Actual Species']=comparison['Actual'].map(dict(enumerate(iris.target_names)))
comparison["Predicted Species"] = comparison["Predicted"].map(dict(enumerate(iris.target_names)))

comparison.head(10)
````

# Evaluate the Model

````python
accuracy = accuracy_score(y_test, y_pred)

print(f"Accuracy: {accuracy:.4f}")
print("\nClassification Report:\n")
print(classification_report(y_test, y_pred,target_names=iris.target_names))
````

````python
cm = confusion_matrix(y_test, y_pred)

disp = ConfusionMatrixDisplay(
    confusion_matrix=cm,
    display_labels=iris.target_names
)

disp.plot()
plt.title("Confusion Matrix")
plt.show()
````

# Visualize the Decision Tree

````python
plt.figure(figsize=(18, 10))

plot_tree(model,feature_names=iris.feature_names,class_names=iris.target_names,filled=True,rounded=True,fontsize=10)
plt.title("Decision Tree Visualization")
plt.show()
````

# Feature Importance

````python

importance_df = pd.DataFrame({
    "Feature": iris.feature_names,
    "Importance": model.feature_importances_
}).sort_values(by="Importance",ascending=False)

importance_df
````

````python

plt.figure(figsize=(9, 5))

plt.bar(importance_df['Feature'],importance_df['Importance'])
plt.title("Feature Importance")
plt.xlabel("Feature")
plt.ylabel("Importance")
plt.xticks(rotation=25)
plt.show()
````

# Compare Gini and Entropy

````python
criteria_results = []

for criterion in ["gini", "entropy"]:
    tree_model = DecisionTreeClassifier(
        criterion=criterion,
        max_depth=3,
        random_state=42
    )
    tree_model.fit(X_train, y_train)

    criteria_results.append({
        "Criterion": criterion,
        "Training Accuracy": tree_model.score(X_train, y_train),
        "Testing Accuracy": tree_model.score(X_test, y_test)
    })

criteria_df = pd.DataFrame(criteria_results)
criteria_df
````

# Effect of max_depth and Overfitting

````python
depth_values = range(1, 11)
train_scores = []
test_scores = []

for depth in depth_values:
    depth_model = DecisionTreeClassifier(
        max_depth=depth,
        random_state=42
    )
    depth_model.fit(X_train, y_train)

    train_scores.append(depth_model.score(X_train, y_train))
    test_scores.append(depth_model.score(X_test, y_test))

depth_results = pd.DataFrame({
    "Max Depth": list(depth_values),
    "Training Accuracy": train_scores,
    "Testing Accuracy": test_scores
})

depth_results    
````

````python
plt.figure(figsize=(9, 5))
plt.plot(depth_values, train_scores, marker="o", label="Training Accuracy")
plt.plot(depth_values, test_scores, marker="o", label="Testing Accuracy")
plt.title("Effect of max_depth on Model Performance")
plt.xlabel("Maximum Tree Depth")
plt.ylabel("Accuracy")
plt.xticks(list(depth_values))
plt.legend()
plt.show()
````

# Predict New Sample

````python
new_flower = pd.DataFrame(
    [[5.9, 3.0, 5.1, 1.8]],
    columns=iris.feature_names
)


predicted_class = model.predict(new_flower)[0]
predicted_species = iris.target_names[predicted_class]

display(new_flower)
print("Predicted class:", predicted_class)
print("Predicted species:", predicted_species)
````

# Display the Decision Path

````python
# Show me every decision you made for this flower.

# Show me every decision you made for this flower.
node_indicator = model.decision_path(new_flower)
# Which leaf node did the flower finally reach?
leaf_id = model.apply(new_flower)[0]

#Which feature is tested at every node?
feature = model.tree_.feature
# The threshold used in every node
threshold = model.tree_.threshold
# gets the first row.
sample = new_flower.iloc[0]

print("Decision path:\n")

for node_id in node_indicator.indices:
    if node_id == leaf_id:
        print(f"Node {node_id}: Leaf node reached.")
        continue

    feature_name = iris.feature_names[feature[node_id]]
    sample_value = sample[feature_name]
    node_threshold = threshold[node_id]
    operator = "<=" if sample_value <= node_threshold else ">"

    print(
        f"Node {node_id}: {feature_name} = {sample_value:.2f} "
        f"{operator} {node_threshold:.2f}"
    )
````
