# K-Nearest Neighbors (KNN) Classification

<details>
<summary><strong>What is KNN Classification?</strong></summary>

K-Nearest Neighbors (KNN) Classification is a **supervised machine learning algorithm** used to predict **categorical class labels**.

Unlike KNN Regression, which predicts a numerical value, KNN Classification predicts the **most common class** among the nearest neighbors.

**Goal:** Classify a new sample based on the majority class of its **K nearest neighbors**.

</details>

---

<details>
<summary><strong>What is K?</strong></summary>

`K` is the **number of nearest neighbors** used to classify a new sample.

Example:

```python
KNeighborsClassifier(n_neighbors=5)
```

The model will consider the **5 closest training samples** before making a prediction.

Choosing the correct value of **K** is important because it affects the model's performance.

</details>

---

<details>
<summary><strong>How Does KNN Classification Work?</strong></summary>

### Step 1
Choose the number of neighbors (**K**).

### Step 2
Calculate the distance between the new sample and every training sample.

### Step 3
Find the **K nearest neighbors**.

### Step 4
Count how many neighbors belong to each class.

### Step 5
Select the class with the **majority vote** so it prefered **k** be **odd** .

### Step 6
Return the predicted class.

</details>

---

<details>
<summary><strong>Example</strong></summary>

Suppose we have the following training data:

| Age | Income | Buys Laptop |
|----:|-------:|:-----------:|
| 22 | 30k | No |
| 25 | 35k | No |
| 28 | 40k | Yes |
| 35 | 60k | Yes |
| 40 | 70k | Yes |

We want to classify a new customer:

```text
Age = 27
Income = 38k
```

Choose:

```text
K = 3
```

Nearest neighbors:

| Age | Income | Class |
|----:|-------:|:-----:|
| 25 | 35k | No |
| 28 | 40k | Yes |
| 22 | 30k | No |

Votes:

```text
No  = 2
Yes = 1
```

Predicted class:

```text
No
```

</details>

---

<details>
<summary><strong>Distance Formula</strong></summary>

KNN uses **Euclidean Distance** by default.

$$
d=\sqrt{(x_1-x_2)^2+(y_1-y_2)^2}
$$

For datasets with more than two features:

$$
d=\sqrt{\sum_{i=1}^{n}(x_i-y_i)^2}
$$

The nearest neighbors are selected based on the smallest distance.

</details>

---

<details>
<summary><strong>Training Process</strong></summary>

```python
from sklearn.neighbors import KNeighborsClassifier

model = KNeighborsClassifier(
    n_neighbors=5
)

model.fit(X_train, y_train)
```

Unlike many machine learning algorithms, KNN has **no actual training phase**.

During `fit()`, the model simply stores the training dataset.

</details>

---

<details>
<summary><strong>Prediction Process</strong></summary>

```python
prediction = model.predict(X_test)
```

During prediction, KNN:

- Computes the distance to every training sample.
- Finds the K nearest neighbors.
- Counts the class labels.
- Uses **majority voting**.
- Returns the predicted class.

</details>

---

<details>
<summary><strong>Choosing the Best K</strong></summary>

A small value of **K**:

- Sensitive to noise.
- Can overfit the data.
- High variance.

A large value of **K**:

- Produces smoother decision boundaries.
- May underfit the data.
- Higher bias.

The best value of **K** is usually selected using **Cross Validation**.

</details>

---

<details>
<summary><strong>Feature Scaling</strong></summary>

KNN is a **distance-based algorithm**.

Features should be scaled before training.

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)
```

Without scaling, features with larger values dominate the distance calculation.

</details>

---

<details>
<summary><strong>Advantages</strong></summary>

- Simple and easy to understand.
- No training phase.
- Easy to implement.
- Works well for small datasets.
- Can model non-linear decision boundaries.

</details>

---

<details>
<summary><strong>Disadvantages</strong></summary>

- Prediction becomes slow for large datasets.
- Requires feature scaling.
- Sensitive to noisy data and outliers.
- Choosing the right value of **K** can be difficult.
- Requires storing the entire training dataset.

</details>

---

<details>
<summary><strong>KNN Classification vs KNN Regression</strong></summary>

| KNN Classification | KNN Regression |
|--------------------|----------------|
| Predicts a class label | Predicts a numerical value |
| Uses majority voting | Uses the mean of neighbors |
| Output is categorical | Output is continuous |
| Used for classification problems | Used for regression problems |

</details>

````python
import pandas as pd
from sklearn import datasets
````

````python
iris=datasets.load_iris()
````

````python
iris
````

````python
print("Features Name  :",iris.feature_names)
````

````python
print("Target Name : ",iris.target_names)
print("Target Values : ",iris.target)
````

````python
df=pd.DataFrame(iris.data,columns=iris.feature_names)
df.head()
````

````python
df['Target']=iris.target
df.head()
````

````python
x=df.drop(['Target'],axis='columns')
y=df['Target']
````

````python
from sklearn.model_selection import train_test_split

x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42,stratify=y)
````

````python
from sklearn.preprocessing import StandardScaler

scaler=StandardScaler()
x_train=scaler.fit_transform(x_train)
x_test=scaler.transform(x_test)
````

````python
from sklearn.neighbors import KNeighborsClassifier

classifier=KNeighborsClassifier(n_neighbors=5)
classifier.fit(x_train,y_train)
````

````python
y_pred=classifier.predict(x_test)
print("Prediction : ",y_pred)
````

````python
from sklearn.metrics import accuracy_score

accuracy=accuracy_score(y_test,y_pred)
print(f"Accuracy :{accuracy*100:.2f}%")
````

````python
from sklearn.metrics import classification_report,confusion_matrix

print(classification_report(y_test,y_pred))
````

````python
cm=confusion_matrix(y_test,y_pred)
print(confusion_matrix(y_test, y_pred))
````

````python
# Visualize Confusion Matrix
import matplotlib.pyplot as plt
from sklearn.metrics import ConfusionMatrixDisplay


ConfusionMatrixDisplay.from_predictions(
    y_test,
    y_pred
)

plt.show()
````
