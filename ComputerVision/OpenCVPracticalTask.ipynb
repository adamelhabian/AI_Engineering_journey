```python
import cv2
import matplotlib.pyplot as plt


# Task 2: Read Image
image = cv2.imread('imgs/newimage.jpg')
cv2.imshow("image",image)

print("Image loaded successfully!")


# Task 3: Print Information
print("\nImage Information")
print("-" * 30)

print("Shape    :", image.shape)
print("Height   :", image.shape[0])
print("Width    :", image.shape[1])
print("Channels :", image.shape[2])
print("Data Type:", image.dtype)
print("Dimensions (ndim):", image.ndim)


# Task 4: Display RGB Image

rgb_image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)

cv2.imshow('rgb_image',rgb_image)


# Task 5: Grayscale
gray_image = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)
cv2.imshow('gray_image', gray_image)



# Task 6: HSV
hsv_image = cv2.cvtColor(image, cv2.COLOR_BGR2HSV)
cv2.imshow("HSV IMAGE : ",hsv_image)
print("\nHSV Shape:", hsv_image.shape)


# Task 7: RGB
rgb_image = cv2.cvtColor(hsv_image, cv2.COLOR_HSV2RGB)
cv2.imshow('rgb_image',rgb_image)



# Task 8: Resize to 300x200
small_image = cv2.resize(image, (300, 200))
print("New Shape:", small_image.shape)


# Task 9: Half Size (INTER_AREA)
# instead of (newW,newH) -> None make opencv calculate the new size automatically

half_image = cv2.resize(
    image,
    None,
    fx=0.5,# half width
    fy=0.5,#half height
    interpolation=cv2.INTER_AREA # the new pixels values are the average of all original pixels
 )


# Task 10: Double Size (INTER_CUBIC)
large_image = cv2.resize(
    image,
    None,
    fx=2,
    fy=2,
    interpolation=cv2.INTER_CUBIC # in enlarging you don't have enough piexels and this attribute calculate needed pixels values by estimating what their color
)

cv2.imshow("Large image",large_image)

# Task 11: Compare Interpolations
nearest = cv2.resize(
    image,
    None,
    fx=2,
    fy=2,
    interpolation=cv2.INTER_NEAREST # copy the nearest pixel
)

cv2.imshow("nearest image ",nearest)

linear = cv2.resize(
    image,
    None,
    fx=2,
    fy=2,
    interpolation=cv2.INTER_LINEAR # use 4 nearby pixels to estimate the new pixels
)
cv2.imshow("linear image ",linear)

cubic = cv2.resize(
    image,
    None,
    fx=2,
    fy=2,
    interpolation=cv2.INTER_CUBIC # use the nearest 16 pixel to estimate the new pixels
)
cv2.imshow("cubic image ",cubic)

# fig, axes = plt.subplots(1, 3, figsize=(15,5))

# axes[0].imshow(cv2.cvtColor(nearest, cv2.COLOR_BGR2RGB))
# axes[0].set_title("INTER_NEAREST")
# axes[0].axis("off")

# axes[1].imshow(cv2.cvtColor(linear, cv2.COLOR_BGR2RGB))
# axes[1].set_title("INTER_LINEAR")
# axes[1].axis("off")

# axes[2].imshow(cv2.cvtColor(cubic, cv2.COLOR_BGR2RGB))
# axes[2].set_title("INTER_CUBIC")
# axes[2].axis("off")

# plt.show()


# Task 12: Save Images
cv2.imwrite("gray_image.jpg", gray_image)
cv2.imwrite("small_image.jpg", small_image)
cv2.imwrite("large_image.jpg", large_image)
print("\nImages Saved Successfully!")


# Task 13: Binary Image
#convert a grayscale image into a black-and-white image.
thersh, binary_image = cv2.threshold(
    gray_image,
    127, # if pixel > 127 change it to 255 (white)
    255,
    cv2.THRESH_BINARY # this the equation  pixel > 127 -> white , else -> black
)

cv2.imwrite("binary_image.jpg", binary_image)

print("Binary image saved successfully!")
cv2.waitKey(0)
```
