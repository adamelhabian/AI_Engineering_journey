## Task 1: Image Thresholding Techniques

## Thresholding Methods Comparison

| Method | Description | Output Behavior | Typical Applications | Advantages | Limitations |
|:-------|:------------|:----------------|:---------------------|:-----------|:------------|
| **THRESH_BINARY** | Converts the image into a black-and-white image. | Pixels > threshold → **255**, Pixels ≤ threshold → **0**. | Object detection, OCR, image segmentation. | Fast, simple, easy to separate objects from the background. | Loses grayscale details and is sensitive to lighting. |
| **THRESH_BINARY_INV** | Inverse of Binary thresholding. | Pixels > threshold → **0**, Pixels ≤ threshold → **255**. | Detecting dark objects, creating inverse masks. | Useful when the object is darker than the background. | Also removes grayscale information. |
| **THRESH_TRUNC** | Limits bright pixel values to the threshold. | Pixels > threshold → **threshold**, Pixels ≤ threshold → **unchanged**. | Brightness control, image preprocessing. | Preserves dark details while reducing bright regions. | Does not create a binary image. |
| **THRESH_TOZERO** | Removes dark pixels and keeps bright pixels. | Pixels > threshold → **unchanged**, Pixels ≤ threshold → **0**. | Highlighting bright objects, feature extraction. | Preserves the original intensity of bright regions. | Dark details are lost. |
| **THRESH_TOZERO_INV** | Removes bright pixels and keeps dark pixels. | Pixels > threshold → **0**, Pixels ≤ threshold → **unchanged**. | Highlighting dark objects, shadow analysis. | Preserves the original intensity of dark regions. | Bright details are lost. |

### Quick Rule

| Method | Easy to Remember |
|:-------|:-----------------|
| **THRESH_BINARY** | Bright → White, Dark → Black |
| **THRESH_BINARY_INV** | Bright → Black, Dark → White |
| **THRESH_TRUNC** | Cap bright pixels at the threshold |
| **THRESH_TOZERO** | Keep only bright pixels |
| **THRESH_TOZERO_INV** | Keep only dark pixels |

```python
# Convert a grayscale image into a binary image using different thresholding methods in OpenCV.

grayimg=cv2.imread('imgs/photo.jpg',0)

# if the pixel val > threshold(127) convert it tto white 255 else convert it black 0 , make it black and white image 
thresh,BinaryImg1=cv2.threshold(grayimg,127,255,cv2.THRESH_BINARY)
# if the pixel val <= threshold (127) convert it to white else covert it to black 0
thresh,BinaryImg2=cv2.threshold(grayimg,127,255,cv2.THRESH_BINARY_INV)
# if the pixel val  > threshold (127) covert it to 127
thresh,BinaryImg3=cv2.threshold(grayimg,127,255,cv2.THRESH_TRUNC)
# if the pixel val > threshold (127)  keep it , if it smaller or equal covert it to 0 black
thresh,BinaryImg4=cv2.threshold(grayimg,127,255,cv2.THRESH_TOZERO)
# it the pixel val > threshold (127) covert it to 0 , if it smaller or equal keep it 
thresh,BinaryImg5=cv2.threshold(grayimg,127,255,cv2.THRESH_TOZERO_INV)


cv2.imshow("Method 1 of binaryImg",BinaryImg1)
cv2.imshow("Method 2 of binaryImg",BinaryImg2)
cv2.imshow("Method 3 of binaryImg",BinaryImg3)
cv2.imshow("Method 4 of binaryImg",BinaryImg4)
cv2.imshow("Method 5 of binaryImg",BinaryImg5)

cv2.waitKey(0)
```

## Task 2: Edge Detection Techniques

## Edge Detection Algorithms Comparison

| Method | Order of Derivative | Edge Detection Approach | Noise Sensitivity | Computational Complexity | Advantages | Disadvantages | Typical Applications |
|:-------|:--------------------|:------------------------|:------------------|:-------------------------|:-----------|:--------------|:---------------------|
| **Laplacian** | **Second-order** | Detects edges by finding rapid intensity changes in all directions. | **High** (very sensitive to noise) | **Low** | Detects edges in every direction, simple to implement. | Produces noisy edges and cannot determine edge direction. | General edge detection, image enhancement, blob detection. |
| **Sobel** | **First-order** | Computes image gradients in the **X** and **Y** directions separately. | **Medium** | **Low** | Detects edge direction, less sensitive to noise than Laplacian. | Less accurate for diagonal edges and fine details. | Object detection, feature extraction, image preprocessing. |
| **Scharr** | **First-order** | Improved Sobel operator using optimized kernels for more accurate gradients. | **Medium-Low** | **Low-Medium** | More accurate than Sobel, especially for small kernels (3×3). | Slightly higher computation than Sobel. | Computer vision, feature extraction, image analysis. |
| **Canny** | **First-order** (uses image gradients) | Multi-stage detector: Gaussian Blur → Gradient → Non-Maximum Suppression → Double Threshold → Edge Tracking. | **Low** | **High** | Produces thin, accurate, and continuous edges with low noise. | More computationally expensive and requires threshold tuning. | Object detection, lane detection, medical imaging, robotics, computer vision. |

### Quick Summary

| Method | Easy to Remember |
|:-------|:-----------------|
| **Laplacian** | Detects **all edges** but is **very sensitive to noise**. |
| **Sobel** | Detects **horizontal and vertical edges** using image gradients. |
| **Scharr** | An **improved Sobel** with more accurate edge detection. |
| **Canny** | The **best overall edge detector**; accurate, thin edges with noise removal. |

```python
# edge detection deal with grayscale image

grayimg=cv2.imread('imgs/photo.jpg',0)

# Detect edges in all direction using second derivative
laplicianIMG=cv2.Laplacian(grayimg,cv2.CV_64F)
cv2.imshow("LaplicianIMG",laplicianIMG)

# use first derivative
# horizontal axes kernal detect vertical direction 
sobleIMGx=cv2.Sobel(grayimg,cv2.CV_64F,1,0)
cv2.imshow("SobelImgx",sobleIMGx)
# vertical aces kernal detect horizontal direction 
sobleIMGy=cv2.Sobel(grayimg,cv2.CV_64F,0,1)
cv2.imshow("SobelImgy",sobleIMGy)

# Imporved version of soble use first derivative 
sharrIMGx=cv2.Scharr(grayimg,cv2.CV_64F,1,0)
cv2.imshow("SharrImgx",sharrIMGx)
# vertical aces kernal detect horizontal direction 
sharrIMGy=cv2.Scharr(grayimg,cv2.CV_64F,0,1)
cv2.imshow("SharrImgy",sharrIMGy)

#pixel > 200 are strong edges , pixel < 100 are noise , what lies between are weak edges , if weak edge connect to strong edge we kept it else we remove it 
cannyIMG=cv2.Canny(grayimg,100,200)
cv2.imshow("CannyIMG",cannyIMG)
cv2.waitKey(0)
```

## Task 3: Harris Corner Detection 

<details>
<summary><strong>Harris Corner Detection (OpenCV)</strong></summary>

## Purpose
Harris Corner Detection is used to **detect corners (interest points)** in an image.  
A corner is a point where **two or more edges meet**. These points are useful for:

- Object detection
- Image matching
- Image stitching (Panorama)
- Feature extraction
- Object tracking


## Results Discussion

- Harris successfully detects **corner points** where two edges intersect.
- The detected corners are highlighted in **red**.
- It is **fast** and works well for images with clear corners.
- It is **rotation invariant**, but **not scale invariant** (zooming the image may reduce accuracy).
- It may also detect extra corners if the image contains **noise**.

---

## Key Functions

| Function | Role |
|----------|------|
| `cv2.cornerHarris()` | Computes the corner response for every pixel. |
| `cv2.dilate()` | Enlarges detected corners to make them easier to see. |
| `corners.max()` | Finds the strongest corner response. |
| `img[corners > 0.01 * corners.max()]` | Selects strong corners using a threshold. |
| `[0,0,255]` | Colors the selected corner pixels **red** (BGR format). |

</details>

```python

# Harris requries float32 image since it make matrix multiplication and produce decimal values
img=cv2.imread("imgs/photo.jpg")
gray = np.float32(grayimg)

# Apply Harris Corner Detection and check if pixel corner or not 
# 2 size of window that we check by on each pixel
# 3 sobel kernel size that used to check image gradient 
# 0.04 sensetive constant control the strict of harris in deciding if the pixel corner or not 
corners = cv2.cornerHarris(gray, 2, 3, 0.04)

# Dilate the corner points to make them more visible increase the affect of corner to make it easy to detect them 
corners = cv2.dilate(corners, None)

# Mark detected corners in red
img[corners > 0.01 * corners.max()] = [0, 0, 255]

# Display images
cv2.imshow("Original Image", cv2.imread("imgs/photo.jpg"))
cv2.imshow("Detected Corners", img)

cv2.waitKey(0)
```
