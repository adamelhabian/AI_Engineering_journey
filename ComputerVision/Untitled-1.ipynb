```python
import cv2
img=cv2.imread('photo.jpg')
cv2.imshow("image",img)

grayimage=cv2.cvtColor(img,cv2.COLOR_BGR2GRAY)
cv2.imshow('gray image',grayimage)

(thresh,BlackAndWhiteImage)=cv2.threshold(grayimage,127,255,cv2.THRESH_BINARY_INV)
cv2.imshow("BlackAndWhite image",BlackAndWhiteImage)
cv2.waitKey(2000)
cv2.destroyAllWindows()
```

```python
import cv2
img=cv2.imread('photo.jpg')

cv2.imshow("o",img)

hsv_img=cv2.cvtColor(img,cv2.COLOR_BGR2HSV)

cv2.imshow("hue",hsv_img[:,:,0])
cv2.imshow("saturation",hsv_img[:,:,1])
cv2.imshow("value",hsv_img[:,:,2])

cv2.waitKey()
cv2.destroyAllWindows()
```
