# Image Preprocessing

Image Preprocessing is the **first step** in Computer Vision.

Its goal is to **improve the quality of an image** before sending it to a Computer Vision model.

**Input:** Original Image  
**Output:** Enhanced Image (better for analysis)

```text
Original Image
      │
      ▼
Image Preprocessing
      │
      ▼
Enhanced Image
      │
      ▼
Computer Vision Model
```

---

## Main Image Preprocessing Techniques

<details>
<summary><strong>Image Segmentation</strong></summary>

Separates an image into meaningful regions or objects.

**Role:** Makes it easier to analyze specific objects.

</details>

<details>
<summary><strong>Data Augmentation</strong></summary>

Creates new training images by applying transformations such as flipping, rotating, or zooming.

**Role:** Increases dataset size and improves model accuracy.

</details>

<details>
<summary><strong>Image Color Model Conversion</strong></summary>

Converts an image from one color space to another (e.g., **BGR → RGB**, **Grayscale**, **HSV**).

**Role:** Uses the most suitable color representation for a task.

</details>

<details>
<summary><strong>Image Resizing</strong></summary>

Changes the image dimensions (width and height).

**Role:** Makes all images the required size for the model.

</details>

<details>
<summary><strong>Image Enhancement</strong></summary>

Improves image quality by reducing noise and increasing visibility.

### Filter Algorithms
- Remove noise.
- Smooth or sharpen the image.

### Contrast Adjusting Techniques
- Increase the difference between dark and bright pixels.
- Make image details easier to see.

</details>

---

## Summary

| Technique | Role |
|-----------|------|
| **Segmentation** | Separate objects from the image. |
| **Data Augmentation** | Create more training images. |
| **Color Model Conversion** | Change the image color space. |
| **Image Resizing** | Resize the image for the model. |
| **Image Enhancement** | Improve image quality. |
| → Filter Algorithms | Remove noise or sharpen details. |
| → Contrast Adjusting | Improve image brightness and contrast. |

# OpenCV Image Reading Flags

<details>
<summary><strong>cv2.IMREAD_COLOR</strong></summary>

- **Default** option.
- Loads the image as a **3-channel BGR** image.
- Ignores the **Alpha (transparency)** channel.
- **Shape:** `(Height, Width, 3)`

</details>

<details>
<summary><strong>cv2.IMREAD_GRAYSCALE</strong></summary>

- Converts the image to **grayscale**.
- Uses **1 channel** (brightness only).
- **Shape:** `(Height, Width)`

</details>

<details>
<summary><strong>cv2.IMREAD_ANYCOLOR</strong></summary>

- Loads the image in its **original color format**.
- Color image → **BGR**
- Grayscale image → **Grayscale**
- Does **not** force color conversion.

</details>

<details>
<summary><strong>cv2.IMREAD_UNCHANGED</strong></summary>

- Loads the image **without any modifications**.
- Preserves **all channels**, including **Alpha (transparency)**.
- Example: PNG with transparency → **BGRA (4 channels)**.

</details>

<details>
<summary><strong>cv2.IMREAD_ANYDEPTH</strong></summary>

- Preserves the image's **original bit depth**.
- Example:
  - **8-bit:** `0–255`
  - **16-bit:** `0–65535`
- Commonly used for **medical** and **scientific** images.

</details>

---

## Summary

| Flag | Channels | Alpha | Bit Depth |
|------|:--------:|:-----:|:---------:|
| `IMREAD_COLOR` | 3 (BGR) | ❌ | 8-bit |
| `IMREAD_GRAYSCALE` | 1 | ❌ | 8-bit |
| `IMREAD_ANYCOLOR` | Original | ❌ | 8-bit |
| `IMREAD_UNCHANGED` | Original | ✅ | Original |
| `IMREAD_ANYDEPTH` | Original | Depends | Original |

# Image Filtering

Image Filtering is an **Image Enhancement** technique used to improve image quality before Computer Vision.

**Goals:**
- Remove noise.
- Smooth the image.
- Sharpen details.
- Detect edges.

```text
Original Image
      │
      ▼
 Image Filter
      │
      ▼
Enhanced Image
```

---

## Types of Filters

<details>
<summary><strong>Low Pass Filter (Smoothing Filter)</strong></summary>

### Role
- Removes **noise**.
- Smooths the image by reducing sharp pixel changes.

### Advantages
- Removes random noise.
- Produces a smoother image.
- Useful before object detection or segmentation.

### Disadvantage
- Makes edges less sharp (blurring).

</details>

<details>
<summary><strong>High Pass Filter (Sharpening Filter)</strong></summary>

### Role
- Highlights **edges** and fine details.
- Keeps high-frequency information.

### Advantages
- Detects object boundaries.
- Sharpens image details.
- Useful for edge detection.

### Disadvantage
- Can also enhance noise.

</details>

---

# Common Blur (Low Pass) Filters

<details>
<summary><strong>1. Averaging Blur (Mean Filter)</strong></summary>

### How it works
- Replaces each pixel with the **average** of its neighboring pixels.

### OpenCV

```python
cv2.blur(img, (3,3))
```

### Advantages
- Very simple.
- Fast.
- Reduces small noise.

### Disadvantage
- Blurs edges significantly.

</details>

<details>
<summary><strong>2. Gaussian Blur</strong></summary>

### How it works
- Uses a **Gaussian (bell-shaped) kernel**.
- Gives **more weight to the center pixel** and less to distant pixels.

### OpenCV

```python
cv2.GaussianBlur(img, (5,5), 0)
```

### Advantages
- Better noise removal than averaging.
- Preserves edges better.
- Most commonly used blur filter.

### Disadvantage
- Still causes slight blurring.

</details>

<details>
<summary><strong>3. Median Blur</strong></summary>

### How it works
- Replaces the center pixel with the **median** value of neighboring pixels.

### OpenCV

```python
cv2.medianBlur(img, 5)
```

### Advantages
- Excellent for **Salt-and-Pepper noise**.
- Preserves edges better than averaging.

### Disadvantage
- Slower than averaging blur.

</details>

<details>
<summary><strong>4. Bilateral Filter</strong></summary>

### How it works
- Smooths the image **while preserving edges**.

### OpenCV

```python
cv2.bilateralFilter(img, 9, 75, 75)
```

### Advantages
- Removes noise.
- Keeps edges sharp.
- Produces high-quality results.

### Disadvantage
- Slowest blur filter.

</details>

---

# Quick Comparison

| Filter | Removes Noise | Preserves Edges | Speed | Best For |
|---------|:-------------:|:---------------:|:-----:|----------|
| Averaging Blur | ✅ | ❌ | ⭐⭐⭐⭐⭐ | Simple smoothing |
| Gaussian Blur | ✅✅ | ✅ | ⭐⭐⭐⭐ | General-purpose blur |
| Median Blur | ✅✅✅ | ✅✅ | ⭐⭐⭐ | Salt & Pepper noise |
| Bilateral Filter | ✅✅ | ✅✅✅ | ⭐ | High-quality smoothing |

---

# Easy Rules to Remember

- **Averaging Blur** → Average neighboring pixels.
- **Gaussian Blur** → Weighted average (center pixel is more important).
- **Median Blur** → Replace with the middle value (best for Salt & Pepper noise).
- **Bilateral Filter** → Remove noise **without blurring edges**.

---

# Summary

| Category | Purpose |
|----------|---------|
| **Low Pass Filter** | Remove noise and smooth the image. |
| **High Pass Filter** | Sharpen details and detect edges. |
| **Averaging Blur** | Fast smoothing using pixel average. |
| **Gaussian Blur** | Smooth image while preserving more details. |
| **Median Blur** | Best for Salt & Pepper noise. |
| **Bilateral Filter** | Smooth image while keeping edges sharp. |

````python
# Image Filtring

import cv2

img=cv2.imread("imgs/photo.jpg")

# Averaging Filter (mean Filter) remove noise and edges
avgblur=cv2.blur(img,(3,3))
cv2.imshow("Mean Filter",avgblur)

#Gaussian Blurring remove gaussian noise
# if we give 0 to sigmax,  both sigma x , y will calculated from kernal size  
Gaussian=cv2.GaussianBlur(img,(7,7),0)
cv2.imshow("Gaussian filter ",Gaussian)

# remove salt and paper noise 
median=cv2.medianBlur(img,5)
cv2.imshow("median filter",median)

#Bilateral blure Remove noise while keep edges sharp
bilateral=cv2.bilateralFilter(img,9,75,75)
cv2.imshow("Bilateral filter",bilateral)
cv2.waitKey(0)
cv2.destroyAllWindows()
````

# Edge Detection Methods

Edge Detection finds the **boundaries of objects** by detecting sudden changes in pixel intensity.

---

<details>
<summary><strong>1. Custom Kernel</strong></summary>

### Derivative
- Depends on the kernel you design.
- Can use **First** or **Second Derivative**.

### Role
- Create your own edge detector for a specific purpose.

### Advantages
- Fully customizable.
- Can detect specific patterns.

### Disadvantages
- Requires manual kernel design.

</details>

---

<details>
<summary><strong>2. Laplacian</strong></summary>

### Derivative
- **Second Derivative**

### Role
- Detects edges in **all directions** (horizontal, vertical, diagonal).

### OpenCV

```python
cv2.Laplacian(img, cv2.CV_64F)
```

### Advantages
- Detects all edge directions.
- Simple implementation.

### Disadvantages
- Very sensitive to noise.
- Usually Gaussian Blur is applied first.

</details>

---

<details>
<summary><strong>3. Sobel</strong></summary>

### Derivative
- **First Derivative**

### Role
- Detects edge direction.
- Uses two gradients:
  - **Gx** → Vertical edges
  - **Gy** → Horizontal edges

### OpenCV

```python
cv2.Sobel(img, cv2.CV_64F, 1, 0)  # Gx
cv2.Sobel(img, cv2.CV_64F, 0, 1)  # Gy
```

### Advantages
- Detects edge direction.
- Less sensitive to noise than Laplacian.
- Fast and widely used.

### Disadvantages
- Less accurate than Scharr for small kernels.

</details>

---

<details>
<summary><strong>4. Scharr</strong></summary>

### Derivative
- **First Derivative**

### Role
- Improved version of Sobel.
- Produces more accurate gradients.

### OpenCV

```python
cv2.Scharr(img, cv2.CV_64F, 1, 0)
```

### Advantages
- More accurate than Sobel.
- Better edge localization.

### Disadvantages
- Slightly slower than Sobel.

</details>

---

<details>
<summary><strong>5. Canny</strong></summary>

### Derivative
- Uses **First Derivative (Sobel)** internally.

### Pipeline
```text
Image
   │
   ▼
Gaussian Blur
   │
   ▼
Sobel Gradient
   │
   ▼
Non-Maximum Suppression
   │
   ▼
Double Threshold
   │
   ▼
Edge Tracking
```

### OpenCV

```python
cv2.Canny(img, 100, 200)
```

### Advantages
- High accuracy.
- Thin and clean edges.
- Removes most false edges.
- Most popular edge detector.

### Disadvantages
- Slower than Sobel.
- Requires choosing two thresholds.

</details>

---

# Quick Comparison

| Method | Derivative | Best For |
|---------|:----------:|----------|
| **Custom Kernel** | Depends | Custom edge detection |
| **Laplacian** | Second | Detecting edges in all directions |
| **Sobel** | First | Horizontal & vertical edge detection |
| **Scharr** | First | More accurate Sobel |
| **Canny** | First (Sobel) | High-quality edge detection |

---

## Easy Rules

```text
Custom Kernel
    ↓
Design your own filter

Laplacian
    ↓
Second Derivative
All edge directions

Sobel
    ↓
First Derivative
Horizontal & Vertical edges

Scharr
    ↓
Better Sobel
More accurate gradients

Canny
    ↓
Gaussian Blur
→ Sobel
→ Thin Edges
→ Threshold
→ Final Edges
```

````python

import cv2
import numpy as np

# Custom Kernal
img = cv2.imread("imgs/photo.jpg", cv2.IMREAD_GRAYSCALE)

kernel = np.array([
    [-1, -1, -1],
    [-1,  8, -1],
    [-1, -1, -1]
])

edges = cv2.filter2D(img, -1, kernel)
cv2.imshow("Edges", edges)

# Laplacin Derivatives
laplacian = cv2.Laplacian(img, cv2.CV_64F)# cv_64F 64 floating point output prevent lossing negative edge info
cv2.imshow("Laplacian", laplacian)

# Sobel Derivatives
# Take first derivative in x direction to detect vertical edges
sobelX = cv2.Sobel(img, cv2.CV_64F, 1, 0)
# Take first derivative in Y direction to detect horizontal edges
sobelY = cv2.Sobel(img, cv2.CV_64F, 0, 1)
cv2.imshow("Sobel X", sobelX)
cv2.imshow("Sobel Y", sobelY)

# Scharr derivatives improved version from sobel
scharrX = cv2.Scharr(img, cv2.CV_64F, 1, 0)
scharrY = cv2.Scharr(img, cv2.CV_64F, 0, 1)
cv2.imshow("Scharr X", scharrX)
cv2.imshow("Scharr Y", scharrY)

edges = cv2.Canny(img, 100, 200)
cv2.imshow("Canny", edges)

cv2.waitKey(0)
cv2.destroyAllWindows()
````

````python
import cv2

img=cv2.imread("imgs/photo.jpg")
gray=cv2.cvtColor(img,cv2.COLOR_BGR2GRAY)

# Create SIFT Detector object
sift=cv2.SIFT_create()
# Find Keypoints in the image search for corners , blobs 
# None -> search the entire image automatic 
kp=sift.detect(gray,None) # list of keypoints each kp cotain info like position , size , angle 

# cv2 draw circles on the copy of img and store the result in img1
img1=cv2.drawKeypoints(gray,kp,img,flags=cv2.DrawMatchesFlags_DRAW_RICH_KEYPOINTS)

# detect keypoint and descripte descriptor (fingerprint for each kp)to each keypoint
kp,des=sift.detectAndCompute(gray,None)
cv2.imshow("img1",img1)

cv2.waitKey(0)

````

````python
# Detect car

import cv2 as cv
carimg=cv.imread('imgs/redcar.jpg')
def nothing ():
    pass
# Create window
cv.namedWindow("TrackBars")
cv.resizeWindow("TrackBars", 640, 240)

# Create Trackbars
cv.createTrackbar("Hue Min", "TrackBars", 0, 179, nothing)
cv.createTrackbar("Hue Max", "TrackBars", 179, 179, nothing)

cv.createTrackbar("Sat Min", "TrackBars", 0, 255, nothing)
cv.createTrackbar("Sat Max", "TrackBars", 255, 255, nothing)

cv.createTrackbar("Val Min", "TrackBars", 0, 255, nothing)
cv.createTrackbar("Val Max", "TrackBars", 255, 255, nothing)

while True :
    #HSV describes  color, intensity, and brightness
    # now opencv can easily answer the qustion like is the pixel is red ?? from hue value 
    #Red = 255 Green = 0  Blue = 0  -> Hue(which color it is 0->179)= Red  Saturation = Strong color  Value= Bright
    # BGR affected by lighting while HSV the hue still indicate same color only birghtiness affected by lighting
    carimg_HSV=cv.cvtColor(carimg,cv.COLOR_BGR2HSV)
    # A mask is simply a black-and-white image
    # we Create Mask to ask each pixel Is your color between lower and upper? 
    # if answer yes make it white 255, it it no make it black 0

    # Get Trackbar values
    h_min = cv.getTrackbarPos("Hue Min", "TrackBars")
    h_max = cv.getTrackbarPos("Hue Max", "TrackBars")

    s_min = cv.getTrackbarPos("Sat Min", "TrackBars")
    s_max = cv.getTrackbarPos("Sat Max", "TrackBars")

    v_min = cv.getTrackbarPos("Val Min", "TrackBars")
    v_max = cv.getTrackbarPos("Val Max", "TrackBars")

    # Create lower and upper arrays
    lower = np.array([h_min, s_min, v_min])
    upper = np.array([h_max, s_max, v_max])



    imgMask=cv.inRange(carimg_HSV,lower,upper)
    imgRes=cv.bitwise_and(carimg,carimg,mask=imgMask) #actually applies the mask to the original image. is the mask white? yes copy the original pixel , no replace it with black

    cv.imshow("red car ",carimg)
    cv.imshow("red car HSV ",carimg_HSV)
    cv.imshow("Mask img of red car ",imgMask)
    cv.imshow("The Result Img ",imgRes)
    
    if cv.waitKey(1) & 0xFF == 27:
        break
cv.destroyAllWindows()    
````

````python
# read Videos
# Video is a group of  series images(frames) 

#This function creates an object that can capture frames from: A webcam A USB camera  A video file
# o default camera ,1 second camera , 2 third camera

cap=cv.VideoCapture('videos/Download (1).mp4')

while(True):
    status,frame=cap.read() # status will be true if there is anther frame to display
    if not status :
        break
    cv.imshow('video',frame)
    # every 1ms show 1 frame
    # i make it each 1 sec show 30 frame
    if cv.waitKey(33) & 0xFF ==ord("q"):break
   


cap.release()# close the video file and frees the resources    
cv.destroyAllWindows() # close every opencv window
````

````python
# Face Detection 
import cv2 

faceCascade=cv2.CascadeClassifier('haarcascade_frontalface_default.xml')
img=cv2.imread('imgs/girl.jpg')
imgGray=cv2.cvtColor(img,cv2.COLOR_BGR2GRAY)

faces=faceCascade.detectMultiScale(imgGray,1.1,4)

for (x,y,w,h)in faces:

    cv2.rectangle(img,(x,y),(x+w,y+h),(0,255,0),3)

cv2.imshow("Face Detection", img)
cv2.waitKey(0)
````
